## Version Control e Github
Github è una piattaforma che permette di visualizzare un intero progetto e tutte le sue versioni. In questo modo, non si perde traccia di ciò che è stato fatto e si può tornare a una versione precedente in caso ci siano problemi di qualsiasi tipo.  
Il version control esiste in diverse forme: locale, il più antico, in cui si aveva tutto su un singolo computer che poteva accedere al progetto e modificarlo; centralizzato, in cui un unico server contiene tutti i file e diversi computer possono accedervi e modificarli; distribuito, in cui diversi computer possono sempre accedere al server ma hanno anche la loro versione locale che possono, una volta apportate le modifiche, caricare al server (è il caso di Git).  
L'esame utilizza git quindi bisogna impararlo. Rizzini non spiega un cazzo, quindi ecco le componenti principali:
- Repository: un indirizzo/cartella che contiene tutti i file del progetto;
- Branch: versioni alternative e temporanee alla Repository, sono una copia su cui si possono applicare determinate modifiche successivamente applicabili al Main della Repository;
- Commit: azione di salvare uno snapshot (un'istantanea) del lavoro fatto;
- Merge: unire alla Repository la Branch selezionata.

### Posizione File e Comandi
I file possono esistere in 4 posti (che Github unisce tutti assieme, ma ragioniamo in locale):
- Working Directory: è la cartella in cui si sta lavorando, in cui esiste il file .git che è il file che controlla la differenza tra la Working Directory e la Local Repository (a sua volta con la Remote Repository). Si inizializza con _git init_;
- Staging Area: un luogo temporaneo in cui immagazzinare i file modificati. Dalla WD vengono trasferiti qui mediante "_git add_" (con l'aggiunta di _-p_, dove p sta per "patch", si possono scandagliare uno per uno i singoli commit da fare per valutare cosa tenere e cosa no);
- Local Repository: è la cartella sul computer su cui si hanno i file della repo principale. Si trasferiscono i file qui mediante "_git commit_";
- Remote Repository: la repo online, per mandarci i file della LR basta usare "_git push_".

Altri comandi utili sono:
- _git fetch_: controlla se la LR è Up To Date con la RR o se ci sono Updates;
- _git pull_: crea una copia locale della RR;
- _git status_: ultime modifiche fatte, i commit fatti...
- _git diff_: vedere le differenze tra due repo.

Una repository in locale deve contenere specifici file:
- .git/: il file che contiene la storia della repo, senza di esso si avrebbe una cartella normale;
- .gitignore: contiene tutti i file che devono essere ignorati nel commit;
- src/: cartella che contiene la parte principale del progetto;
- README.md: contiene le istruzioni e informazioni del progetto, assieme a gitignore è obbligatorio al first commit.

### Pull Request e Merge Request
Una volta che si sono apportate delle modifiche rilevanti, si possono effettuare due diverse operazioni: **Pull Request** e **Merge Request**.  
Il **Pull request** non è un'azione di Github, bensì una sovrastruttura di Github che permette ai diversi developer di analizzare la richiesta di unire la main repository con il branch su cui si è lavorato.  
Se quest'analisi passa, si effettua una **Merge Request** ovvero l'unione della Local Repository alla Remote Repository.  
Mentre la PR si effettua da Github, comandi fondamentali per gestire le branches sono:
- _git switch_: seleziona la branch in cui si vuole lavorare, e se si vuole creare una nuova repo è necessario utilizzare _git switch -c_
- _git merge_: unisce la branch alla main.

### Altro
Si può lavorare o su una copia non condivisa della main, oppure sulla copia che è attualmente condivisa con il team. Nel primo caso si usa _git fork_, nel secondo _git clone_.  
Con le funzioni di _checkout_ si può tornare a versioni precedenti.  
Per fare add di più file: _git add ._
