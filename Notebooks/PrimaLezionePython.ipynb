```python
# With n samples, with sample rate SR: T = n/SR. f = 1/T. 
amplitude = 1
SR = 48000
n = round(SR/440)
saw_period = [
    amplitude * (2 * x / n - 1)
    for x in range(n)
]
saw = saw_period * 1600
Audio(saw, rate=SR, normalize=False)
# To get 440Hz, we can do n = round(SR/440). n must be integer!
```

```python
half_n = 150
up = [-amplitude + 2 * amplitude * x / half_n
      for x in range(half_n)]
down = [amplitude - 2 * amplitude * x / half_n
        for x in range(half_n)]
full_triangle = up + down
```

```python
Audio(full_triangle * 1000, rate = 96000, normalize = False)
```
