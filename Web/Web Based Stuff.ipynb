# Web based stuff
There is a paradigm [WYSIWYG](https://it.wikipedia.org/wiki/WYSIWYG) ("What you see is what you get"), where what you write can get exactly written onto the page you want to print.  
Latex is not WYSIWYG, Word is.  
Codepen has on the right the output of what you've written on the left right after.

## Markup languages for web programming
HTML is a markup language (HyperText Markup Language).  
What you write in index.html will go into the output, in style.css and script.js it doesn't.  
In the html file you can put the content and the structure (paragraph, heading...).

In CSS (Cascading Style Sheets) you modify the appearance (colors, styles, shapes, sizes...).

In JavaScript you can program what the web does when it interfaces with the user (?)

### HTML
- This notation indicates a paragraph `<p></p>`
- This means anchor and it's used for links `<a></a>`. 
- This means bold `<b></b>`
- This means section `<div></div>`
- There are extremely many, you should look at a cheat sheet online.

### CSS
- target is the type of content which will be affected by the rule
- parameter is what exact characteristic of the content will be affected
- value is the amount that will be assigned to it
